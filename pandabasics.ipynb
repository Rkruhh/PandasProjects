```python
# Panda Series is a one-dimensional array holding data of any type

import pandas as pd
a = [1,2,3,4]
var = pd.Series(a)
print(var)
print(var[0])
```

```python
import pandas as pd
df = pd.read_excel("exercise_data.xlsx")
new_df = df.dropna()
print(new_df.to_string())
```

```python
import pandas as pd
df = pd.read_excel("exercise_data.xlsx")
df.fillna(130, inplace=True)
print(df.to_string())
```

```python
import pandas as pd
df = pd.read_excel("exercise_data.xlsx")
x = df['Calories'].mean()
df.fillna({"Calories": x}, inplace=True)
print(df.to_string())
```

```python
import pandas as pd
df = pd.read_excel("exercise_data.xlsx")
df['Date'] = pd.to_datetime(df['Date'], format="mixed")
print(df.to_string())
```

```python
import pandas as pd
df = pd.read_excel("exercise_data.xlsx")
df.loc[7, 'Duration'] = 45
print(df.to_string())
```

```python
import pandas as pd
df = pd.read_excel("exercise_data.xlsx")

for x in df.index:
  if df.loc[x, 'Duration'] > 120:
    df.loc[x, 'Duration'] = 120
    print(df.to_string())
```

```python
import pandas as pd
df = pd.read_excel("exercise_data.xlsx")
df.drop_duplicates(inplace=True)
print(df)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_excel("exercise_data.xlsx")
df.plot()
plt.show()
```

```python
# Scatter plot

import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_excel("exercise_data.xlsx")

df.plot(kind = 'scatter', x = 'Duration', y = 'Calories')

plt.show()
```
